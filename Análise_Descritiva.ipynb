### Para avaliarmos o comportamento da variável RENDA vamos construir uma tabela de frequências considerando as seguintes classes em salários mínimos (SM)
#### <font color='blue'>Descreva os pontos mais relevantes que você observa na tabela e no gráfico.</font>

Classes de renda:

<b>A</b> ► Acima de 25 SM

<b>B</b> ► De 15 a 25 SM

<b>C</b> ► De 5 a 15 SM

<b>D</b> ► De 2 a 5 SM

<b>E</b> ► Até 2 SM

Para construir as classes de renda considere que o salário mínimo na época da pesquisa era de <b>R$ 788,00</b>.

#### Siga os passos abaixo:

### 1º Definir os intevalos das classes em reais (R$)

```python
classes = [dados.Renda.min(), 1576, 3940, 11820, 19700, dados.Renda.max()]
classes
```

### 2º Definir os labels das classes

```python
labels = ['E', 'D', 'C', 'B', 'A']
```

### 3º Construir a coluna de frequências

```python
frequencias = pd.value_counts(pd.cut(x = dados.Renda,
                                     bins = classes,
                                     labels = labels,
                                     include_lowest = True))
frequencias
```

### 4º Construir a coluna de percentuais

```python
percentual = pd.value_counts(pd.cut(x = dados.Renda,
                                     bins = classes,
                                     labels = labels,
                                     include_lowest = True),normalize = True) * 100
percentual
```

### 5º Juntar as colunas de frequência e percentuais e ordenar as linhas de acordo com os labels das classes

```python
freq_por = pd.DataFrame({'Frequência': frequencias,
                               'Porcentagem (%)': percentual})
freq_por.rename_axis('Classes', axis= 'columns', inplace = True)
freq_por.sort_index(ascending = False)
```

### Construa um gráfico de barras para visualizar as informações da tabela de frequências acima

```python
# ax = sns.barplot(x = labels, y=freq_por['Porcentagem (%)'],
#                  color = 'blue', saturation = .5)
freq_por['Frequência'].plot.bar(width = 1, color = 'blue', alpha = 0.6, figsize=(14, 6))
```

> ### Conclusões

Com este gráfico é visivel que a distribuição de renda está muito desigual pois apenas 1% da população recebe mais de 15 salarios minimos e 0,5% da população recebe mais de 25

### Obtenha a média, mediana e valor máximo da variável RENDA segundo SEXO e COR
#### <font color='blue'>Destaque os pontos mais importante que você observa nas tabulações</font>
#### <font color='red'>O parâmento <i>aggfunc</i> da função <i>crosstab()</i> pode receber uma lista de funções. Exemplo: <i>aggfunc = {'mean', 'median', 'max'}</i></font>

```python
percentual = pd.crosstab(dados.Cor,
                         dados.Sexo,
                         aggfunc = ['mean', 'median', 'max'],
                         values = dados.Renda)
percentual.rename(columns = sexo, inplace = True)
percentual.rename(index = cor, inplace = True)

percentual
```
